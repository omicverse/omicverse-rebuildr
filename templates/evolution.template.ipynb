## Setup

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
PORT = Path('..').resolve()
sys.path.insert(0, str(PORT))

# History — appended to once per iteration
history = []  # list of dict: {iter, title, wall_clock_s, parity, status, narrative}

def _record(idx, title, wall_clock, parity, status, narrative):
    history.append({'iter': idx, 'title': title, 'wall_clock_s': wall_clock,
                    'parity': parity, 'status': status, 'narrative': narrative})

def _plot(threshold=None):
    fig, (a1, a2) = plt.subplots(2, 1, figsize=(7, 6), sharex=True)
    its = [h['iter'] for h in history]
    a1.plot(its, [h['wall_clock_s'] for h in history], 'o-', color='#0078d4')
    a1.set_ylabel('wall-clock (s)'); a1.set_yscale('log'); a1.set_title('time')
    a2.plot(its, [h['parity'] for h in history], 'o-', color='#a4262c')
    if threshold is not None:
        a2.axhline(threshold, ls='--', color='red', alpha=0.6, label=f'threshold {threshold}')
        a2.legend()
    a2.set_xlabel('iteration'); a2.set_ylabel('parity metric'); a2.set_title('parity')
    plt.tight_layout(); plt.show()
```
